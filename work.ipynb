techniques for filling missing values by using fillna()

```python
df['Attendance(%)'] = df['Attendance(%)'].fillna(df['Attendance(%)'].median())
```

```python
print(df["Attendance(%)"])
```

```python
df['Study hours/week'] = df['Study hours/week'].fillna(df['Study hours/week'].median())
```
